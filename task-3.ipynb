```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

model_name = "finetuned_model" 
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)

model.eval()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
```

```python
import shap
import datasets
from transformers import pipeline
model.config.id2label = {0: 'Human', 1: 'AI'}
model.config.label2id = {'Human': 0, 'AI': 1}
pipe = pipeline("text-classification", model=model, tokenizer=tokenizer, top_k=None, device=0 if torch.cuda.is_available() else -1)

explainer = shap.Explainer(pipe)
```

```python
test_text = None
with open("books/crime_and_punishment.txt", "r") as file:
    test_text = file.read()
    test_text = test_text.split("\n\n")

results = pipe(test_text, truncation=True, batch_size=8)
ai_paragraphs = []

for text, prediction in zip(test_text, results):
    top_label = max(prediction, key=lambda x: x['score'])['label']
    if top_label == 'AI':
        ai_paragraphs.append(text)

test_text =  datasets.Dataset.from_dict({"text": ai_paragraphs})
```

```python
shap_values = explainer(test_text["text"])
shap.plots.text(shap_values)
```
